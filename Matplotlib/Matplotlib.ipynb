```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

#### **Matplotlib Basics**

```python
x = np.linspace(0,5,11)
x
```

```python
y = x**2
y
```

```python
plt.plot(x,y)
plt.title("Hello")
plt.xlabel("X axis")
plt.ylabel("Y axis")
```

#### **Subplots in Matplotlib**

```python
plt.subplot(2,2,1)
plt.plot(x**2,y)
plt.subplot(2,2,3)
plt.plot(x,y)
plt.subplot(2,2,4)
plt.plot(y**2,x)
plt.subplot(2,2,2)
plt.plot(y,x)
```

#### **Object Oriented Way**

```python
fig = plt.figure()
axes1 = fig.add_axes([.1,.3,1,1])
axes1.plot(x,y)
axes2 = fig.add_axes([0.3,0.4,.3,.3])
axes2.plot(x,y)
```

#### **Type of Plots**

```python
plt.scatter(x,y)
```

```python
plt.hist(x)
```

```python
plt.boxplot(x)
```

#### **Saving Plots**

```python
fig = plt.figure()
ax = fig.add_axes([.3,.3,.8,.8])
ax.plot(x,y)
plt.savefig("Basicplot.png")
```
