### **Loading the Libraries and Dataset**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings("ignore")

df = pd.read_csv("IPL.csv")
```

### **Basic Information**

```python
df.info()
```

**Check the size of rows and columns of the Dataset**

```python
print(f"Your Rows are {df.shape[0]} and columns are {df.shape[1]}")
```

**Now let's see how many columns have null values in total.**

```python
df.isnull().sum()
```

### **Now, Here comes some Basic Questions**

**1. Which team won the most matches?**

```python
match_wins = df["match_winner"].value_counts()
sns.barplot(x=match_wins.index, y=match_wins.values, palette="viridis")
plt.title("Most Match Win by Team")
plt.xticks(rotation=90)
plt.xlabel("Teams")
plt.ylabel("Matches")
```

**2. Toss Decision Trends**

```python
toss_decision_trends = df["toss_decision"].value_counts()
sns.barplot(x= toss_decision_trends.index, y= toss_decision_trends.values, palette="deep")
plt.title("Toss Decision Trends")
plt.xlabel("Toss Options")
plt.ylabel("Number od Times")
```

**3. Toss Winner vs Match Winner**

```python
countNum = df[df["toss_winner"] == df["match_winner"]]["match_id"].count()
countPercentage = (countNum * 100)/df.shape[0]
countPercentage.round(2)
```

**4. How do teams win?(Runs vs Wicket)**

```python
sns.countplot(x=df["won_by"], palette="viridis")
plt.title("Teams win by Runs or Wicket")
```

### **Key Player Performances**

**1.  Most "Player of the match" awards**

```python
count_value = df["player_of_the_match"].value_counts().head(5)
plt.figure(figsize=(7.5,4.8))
sns.barplot(x=count_value.index, y=count_value.values, palette="mako")
plt.title("Top 5 Players with man of the match")
plt.xlabel("")
```

**2. Top Scorers**

```python
top_scorer = df.groupby("top_scorer")["highscore"].sum().sort_values(ascending=False).head()
plt.figure(figsize=(7.5,4.8))
sns.barplot(x=top_scorer.index,y=top_scorer.values,palette="viridis")
plt.title("Top Scorers")
plt.xlabel(' ')
plt.tight_layout()
plt.savefig("Top Scorer.png")

```

**10 Best Bowling Figures**

```python
df["highest_wickets"] = df["best_bowling_figure"].apply(lambda x:x.split("--")[0])
df["highest_wickets"] = df["highest_wickets"].astype(int)
top_bowlers = df.groupby("best_bowling")["highest_wickets"].sum().sort_values(ascending=False).head(10)
plt.figure(figsize=(7.5,4.8))
sns.barplot(x=top_bowlers.values, y=top_bowlers.index, palette="deep")
plt.title("Best Bowling Figures")
plt.ylabel("")
```

#### **Venue Anaylsis**

**Most matches played by venue**

```python
high_venue = df["venue"].value_counts()
sns.barplot(x=high_venue.values, y=high_venue.index, palette="mako")
plt.title("Most Matches Played by venue")
plt.ylabel("")
```
