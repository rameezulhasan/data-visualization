```python
import seaborn as sns
import matplotlib.pyplot as plt
```

We use distribution plots to visualize the distribution of quantitative data.
Let's discuss some plots that allows us to visualize the distribution of a data set. These plots are: 

- **distplots/histplot:**  is main ye hota ke aik bin ya box banta hai different values ke darmayan aur jo values us ke andar hongi wo us bin main add ho jaye ga. jaise bin banay ho 0-10, 10-20, 20-30 etc to jo values jahan ati hai wo us bin main add ho jaye gi
- jointplot
- pairplot
- rugplot

```python
df = sns.load_dataset("tips")
df
```

```python
df["size"].unique()
```

```python

sns.displot(df["total_bill"], kde= True)
```

### **Join Plot**

```python
sns.jointplot(x= "total_bill", y="tip", data = df, kind="scatter")
```

```python
sns.jointplot(x= "total_bill", y="tip", data = df, kind="reg")
```

### **PairPlot**

```python
sns.pairplot(df, hue="sex")
```

```python
sns.pairplot(df, hue="sex", palette="rainbow")
```
